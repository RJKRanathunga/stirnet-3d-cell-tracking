# Stage 3 instance-segmentation analysis

This launcher installs the reusable Stage 3 diagnostic widget. Use Napari's native T slider (including normal Ctrl+scroll navigation) to select a scene time index; the widget shows its mapped original frame and loads only that frame's saved Stage 6 data. Stage 3 runs only when **Run analysis** is pressed.

```python
import napari

from src.io import PipelinePaths
from diagnostics.stage_analysis import add_stage3_analysis_widget
```

```python
paths = PipelinePaths.discover()
```

```python
viewer = napari.Viewer(ndisplay=3)
stage3_widget = add_stage3_analysis_widget(viewer, paths=paths)
```

The widget enforces a 3D view after scene, time, layer, parameter, peak-setting, and hypothesis changes. All temporary configuration changes remain in memory and all diagnostic layers are tagged with `metadata["owner"] = "stage3-analysis"`.
